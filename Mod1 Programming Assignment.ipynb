```python
print("How many seconds are in an hour?")

60*60
sec_per_hour = 60*60
print(sec_per_hour)
print("How many seconds are in one day?")
sec_per_hour * 24
sec_per_day = sec_per_hour * 24
print(sec_per_day)
print("What is the result of dividing seconds_per_day by seconds_per_hour using floating-point division?")
print(sec_per_day / sec_per_hour)
print("What is the result of dividing seconds_per_day by seconds_per_hour using integer division?")
print(sec_per_day // sec_per_hour)
```
